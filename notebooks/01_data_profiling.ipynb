```python
import os

BASE = '/content/drive/MyDrive/nyc-capital-projects-analysis'
RAW = f'{BASE}/data/raw'
PROCESSED = f'{BASE}/data/processed'

print(os.listdir(RAW))
```

```python
import pandas as pd

files = ['budget_schedule', 'schedule_history', 'budget_spend_history']
dfs = {name: pd.read_csv(f'{RAW}/{name}.csv', low_memory=False) for name in files}

for name, df in dfs.items():
    print(f'\n===== {name} =====')
    print(f'Rows: {df.shape[0]:,}  |  Columns: {df.shape[1]}')
    print(df.dtypes.to_string())
```

```python
bs = dfs['budget_schedule']
sh = dfs['schedule_history']
bsh = dfs['budget_spend_history']

print('Reporting periods in budget_schedule:\n', bs['Reporting Period'].value_counts().sort_index().to_string())
print('\nUnique FMS IDs:', bs['FMS ID'].nunique(), '| Unique PIDs:', bs['PID'].nunique())
print('Rows with no PID:', bs['PID'].isna().sum())

print('\nCurrent Phase:\n', bs['Current Phase'].value_counts(dropna=False).to_string())
print('\nCompletion Date Type:\n', sh['Completion Date Type'].value_counts(dropna=False).to_string())
print('\nTop 15 reasons for completion change:\n', sh['Reason for Forecast Completion Change'].value_counts().head(15).to_string())
print('\nManaging agencies:\n', bs['Managing Agency'].value_counts().to_string())
```
